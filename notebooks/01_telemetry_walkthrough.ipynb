# F1 telemetry walkthrough

An interactive walkthrough of the analysis. Set up the environment as described in `README.md` first. The notebook calls the reusable functions in `src/f1_telemetry/` rather than duplicating the logic.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from f1_telemetry.analysis import load_fastest_laps, align_telemetry, calculate_metrics, plot_comparison
```

```python
drivers = ('VER', 'PER')
session, telemetry, lap_info = load_fastest_laps(2024, 'Bahrain', 'R', drivers, ROOT / 'cache')
aligned = align_telemetry(telemetry[drivers[0]], telemetry[drivers[1]])
metrics = calculate_metrics(aligned, *drivers)
metrics.round(3)
```

```python
plot_comparison(aligned, drivers, ROOT / 'outputs' / 'telemetry_comparison.png')
from IPython.display import Image, display
display(Image(filename=ROOT / 'outputs' / 'telemetry_comparison.png'))
```

## Observations

1. **Time is lost mostly in the corners.** Top speeds are nearly identical (VER 301.0 km/h, PER 302.8 km/h), yet the delta grows in steps at each corner. Corner performance, not straight-line speed, drives the 1.589 s gap.
2. **Earlier throttle application is associated with the gains.** VER is at full throttle for 62.3% of the lap compared with 58.5% for PER. On several corner exits (e.g. Turn 1, ~700–900 m) he returns to the throttle earlier and carries more speed onto the next straight.
3. **Higher minimum corner speed in the medium-speed section.** Around 1,800–2,000 m, VER's minimum speed is about 15–20 km/h higher than PER's, and the delta rises sharply there.

These are associations in a single-lap comparison. Tyre condition, fuel load, traffic and energy deployment were not controlled for.
